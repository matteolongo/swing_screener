# 16 — Analysis Module Quickstart

Volume-zone analysis (pure advisory, no state mutation). See the [analysis README](https://github.com/matteolongo/swing_screener/blob/main/src/swing_screener/analysis/README.md) for full documentation.

```python
from __future__ import annotations
import pandas as pd

pd.set_option("display.width", 140)

from swing_screener.data.providers import get_market_data_provider
from swing_screener.analysis.volume_zones import analyze_volume_zones, VolumeZoneConfig
```

```python
provider = get_market_data_provider()
ohlcv = provider.fetch_ohlcv(["AAPL"], "2024-01-01", "2024-12-31")
```

```python
result = analyze_volume_zones("AAPL", ohlcv, lookback=120, min_rr=2.0)
print(f"Profile type: {result.profile_type}")
print(f"Setup: {result.setup_type}, Action: {result.action}")
rr_val = result.trade_plan.rr
rr_str = f"{rr_val:.2f}" if rr_val is not None else "None"
print(f"Entry: {result.trade_plan.entry}, Stop: {result.trade_plan.stop}, Target: {result.trade_plan.target}, RR: {rr_str}")
print(f"Market bias: {result.market_bias}, Confidence: {result.confidence_score}")
print(f"Key levels: price={result.key_levels.price}, poc={result.key_levels.poc}, vwap={result.key_levels.vwap}")
```

```python
for zone in result.volume_zones:
    print(f"  {zone.role}: {zone.price_low:.2f} - {zone.price_high:.2f} ({zone.kind.upper()})")
```
